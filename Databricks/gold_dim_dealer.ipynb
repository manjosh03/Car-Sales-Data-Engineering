# CREATING DIMENSIONS Dealer

### Fetch Relative Columns

```python
%sql
select * from parquet.`abfss://silver@mbcarsdatalake.dfs.core.windows.net/car_sales`
```

```python

df_src = spark.sql('''
select distinct(Dealer_ID) as Dealer_ID, DealerName
from parquet.`abfss://silver@mbcarsdatalake.dfs.core.windows.net/car_sales`
''')
```

```python
df_src.display()
```

### Dim_Dealer Sink - Initial and Incremental (Just bring the schema if table does not exists)

```python
if spark.catalog.tableExists('cars_catalog.gold.dim_dealer'):
    df_sink = spark.sql('''
    SELECT dim_dealer_key, Dealer_ID, DealerName 
    from cars_catalog.gold.dim_dealer
    ''')

else:    
    df_sink = spark.sql('''
    SELECT  1 as dim_dealer_key, Dealer_ID, DealerName 
    from parquet.`abfss://silver@mbcarsdatalake.dfs.core.windows.net/car_sales`
    WHERE 1=0
    ''')
```

### Filtering new records and old records 

```python
df_filter = df_src.join(df_sink, df_src['Dealer_ID'] == df_sink['Dealer_ID'], 'left').select(df_src['Dealer_ID'], df_src['DealerName'], df_sink['dim_dealer_key'])
df_filter.display()
```

**df_filter_old**

```python
df_filter_old = df_filter.filter(col("dim_dealer_key").isNotNull())
```

```python
df_filter_old.display()
```

**df_filter_new**

```python
df_filter_new = df_filter.filter(col("dim_dealer_key").isNull()).select(df_src['Dealer_ID'], df_src['DealerName'])    
```

```python
df_filter_new.display()
```

### Create Surrogate Key

***Fetch the max Surrogate Key from existing table***

```python
if (incremental_flag == '0'):
    max_value = 1 
else:
    max_value_df = spark.sql("SELECT MAX(dim_dealer_key) FROM cars_catalog.gold.dim_dealer")
    max_value = max_value_df.collect()[0][0]+1

```

**Create Surrogate key column and ADD the max**

```python
df_filter_new = df_filter_new.withColumn('dim_dealer_key', max_value + monotonically_increasing_id())
```

```python
df_filter_new.display()
```

### Create Final DF = df_filter_old + df_filter_new

```python
df_final = df_filter_new.union(df_filter_old)
df_final.display()
```

# SCD TYPE - 1 (UPSERT)

```python
from delta.tables import DeltaTable
```

```python
#Incremental Run
if spark.catalog.tableExists('cars_catalog.gold.dim_dealer'):
    delta_tbl = DeltaTable.forPath(spark, "abfss://gold@mbcarsdatalake.dfs.core.windows.net/dim_dealer")
    delta_tbl.alias("trg").merge(
        df_final.alias("src"),
        "trg.dim_dealer_key = src.dim_dealer_key")\
            .whenMatchedUpdateAll()\
                .whenNotMatchedInsertAll()\
                    .execute()

#Initial Run
else:
    df_final.write.format("delta")\
        .mode("overwrite")\
            .option("path", "abfss://gold@mbcarsdatalake.dfs.core.windows.net/dim_dealer")\
                .saveAsTable("cars_catalog.gold.dim_dealer")
```
